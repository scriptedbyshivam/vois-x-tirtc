### **Univariate Analysis: Income, Visits, and Illness Distributions**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# 1. Distribution of Income
sns.histplot(df['income'], bins=15, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Distribution of Income')
axes[0].set_xlabel('Income')
axes[0].set_ylabel('Frequency')

# 2. Count of Doctor Visits
sns.countplot(x='visits', data=df, ax=axes[1], palette='crest', hue='visits', legend=False)
axes[1].set_title('Frequency of Doctor Visits')
axes[1].set_xlabel('Number of Visits')
axes[1].set_ylabel('Count')

# 3. Count of Illnesses
sns.countplot(x='illness', data=df, ax=axes[2], palette='flare', hue='illness', legend=False)
axes[2].set_title('Distribution of Number of Illnesses')
axes[2].set_xlabel('Number of Illnesses')
axes[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

**10. Visualize the percentage of people getting govt health insurance due to low income, due to old age and also the percentage of people having private health insurance.**

```python
import matplotlib.pyplot as plt

# 0. Common labels for the pie charts
label = ['yes', 'no']

# Create a single figure with 1 row and 3 columns of subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# --- Chart 1: Govt Insurance due to Low Income ---
Y1 = df[df['freepoor'] == 'yes']
N1 = df[df['freepoor'] == 'no']
x1 = [Y1.shape[0], N1.shape[0]]

axes[0].pie(x1, labels=label, autopct='%1.1f%%', startangle=90)
axes[0].set_title("% of people getting govt health \nInsurance due to low income")

# --- Chart 2: Private Insurance ---
Y2 = df[df['private'] == 'yes']
N2 = df[df['private'] == 'no']
x2 = [Y2.shape[0], N2.shape[0]]

axes[1].pie(x2, labels=label, autopct='%1.1f%%', startangle=90)
axes[1].set_title("% of people having private Insurance")

# --- Chart 3: Govt Insurance due to Old Age, Disability, or Veteran Status ---
Y3 = df[df['freerepat'] == 'yes']
N3 = df[df['freerepat'] == 'no']
x3 = [Y3.shape[0], N3.shape[0]]

axes[2].pie(x3, labels=label, autopct='%1.1f%%', startangle=90)
axes[2].set_title("% of people getting govt health Insurance \ndue to old age, disability or veteran status")

# Adjust spacing so titles do not overlap
plt.tight_layout()
plt.show()
```

```python
# % of people getting govt insurance due to low income
label=['yes','no']
Y = df[df['freepoor']=='yes']
N = df[df['freepoor']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(7,7))
plt.pie(x,labels=label)
plt.title("% of people getting govt health insurance  ")
plt.show()
# % of people having private insurance
Y = df[df['private']=='yes']
N = df[df['private']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(7,7))
plt.pie(x,labels=label)
plt.title("% of people having private health insurance  ")
plt.show()
# % of people getting govt innsurance due to old age, disability or veteran status
Y = df[df['freerepat']=='yes']
N = df[df['freerepat']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(7,7))
plt.pie(x,labels=label)
plt.title("% of people getting govt health insurance due to old age, disability or veteran status ")
plt.show()
```

**11. Plot a horizontal bar chart to analyze the reduce days od activity due to illness based on Gender**

```python
db= df.groupby('gender')['reduced'].sum().to_frame().reset_index()
#Creating the bar chart
plt.barh(db['gender'],db['reduced'],color = ['cornflowerblue','lightseagreen'])
#Adding the aesthetics
plt.title('Bar Chart')
plt.xlabel('gender')
plt.ylabel('reduced activity')
#show thw plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Aggregate the total reduced activity days by gender
reduced_by_gender = df.groupby('gender')['reduced'].sum().reset_index()

# Create a bar plot
plt.figure(figsize=(8, 6))
sns.barplot(x='gender', y='reduced', data=reduced_by_gender, palette='Set2', hue='gender', legend=False)

# Add aesthetics
plt.title('Total Reduced Activity Days by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Total Days of Reduced Activity', fontsize=12)

# Annotate bars with values
for index, row in reduced_by_gender.iterrows():
    plt.text(index, row['reduced'] + 50, str(int(row['reduced'])), color='black', ha="center", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate correlation matrix for numeric columns only
corr_matrix = df.corr(numeric_only=True)

# Set up the matplotlib figure
plt.figure(figsize=(10, 8))

# Draw the heatmap with a nice color palette and annotations
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

# Add aesthetics
plt.title('Correlation Heatmap of Numerical Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Calculate the average reduced activity days by gender
mean_reduced_by_gender = df.groupby('gender')['reduced'].mean().reset_index()
display(mean_reduced_by_gender)
```

```python
illness_analysis = (
    df.groupby("illness")["visits"]
    .agg(["count", "mean"])
    .reset_index()
)

display(illness_analysis)

sns.lineplot(
    data=illness_analysis,
    x="illness",
    y="mean",
    marker="o"
)

plt.title("Illness score vs. Average Doctor visits")
plt.xlabel("illness score")
plt.ylabel("Mean recorded visits")
plt.show()
```
